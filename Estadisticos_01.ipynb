```python
# Por individuo, cuántas clases llevaste el semestre pasado?
numero_de_clases = np.array([4,6,6,6,4,5,5,5,6,4,5,5,3,4,5,6])

4+6+6+6+4+5+5+5+6+4+5+5+3+4+5+6
```

Fórmula del **promedio arítimetico**
$$\bar{X}=\frac{\sum_{i=1}^nx_i}{n}.$$

Notación:
* $\Sigma:$ "sigma" letra del alfabeto griego.
* $\sum_{i=1}^nx_i= x_1+x_2+x_3+\dots+x_n$


```python
# Calcular la suma de los datos
suma=np.sum(numero_de_clases)
print('La sumatoria es',suma)
n=len(numero_de_clases)
# El promedio de numero de clase es
promedio=suma/n
print('El promedio es',promedio)
79/16
```

```python
import statistics as sts
# sts.mean()  calcula la MEDIANA
# np.mean()   calcula el promedio
Avr0=np.mean(numero_de_clases)
print('El promedio es',Avr0)

```

```python
plt.hist(numero_de_clases, edgecolor='black', alpha=0.7, label = 'histograma')
plt.axvline(Avr0, color='r', linestyle='dashed', linewidth=2.0, label = 'media')
plt.legend()
plt.xlabel(r'Numero de cursos $X$')
plt.ylabel(r'Frecuencia $f$')
plt.title('Numero de cursos por alumno')
plt.show()
```

# **Desviación Estandar**

##$$
\sigma^2 = \frac{\sum_{i=1}^{n} \big[ f_i\ \cdot (X_i - \bar{X})^2) \big]}{n-1}
$$

# **Varianza**

##$$
\sigma = \sqrt{\frac{\sum_{i=1}^{n} \big[ f_i\ \cdot (X_i - \bar{X})^2) \big]}{n-1}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

f = np.array([1,4,6,5])
X = np.array([3,4,5,6])

numerador = np.sum(f*(X - Avr0)**2)
denominador = np.sum(f) - 1

sigma2=numerador/denominador
sigma=np.sqrt(numerador/denominador)


print('La desviacion estandar es', np.round(sigma2,2))
print('La varianza es', np.round(sigma,2))
```

```python
plt.hist(numero_de_clases, edgecolor='black', alpha=0.7, label = 'Histograma')
plt.axvline(Avr0, color='r', linestyle='dashed', linewidth=2.0, label = r'$\bar{X}$')
plt.axvline(Avr0+sigma, color='k', linestyle='dashed', linewidth=2.0, label = r'$\bar{X}+\sigma$')
plt.axvline(Avr0-sigma, color='k', linestyle='dashed', linewidth=2.0, label = r'$\bar{X}-\sigma$')
plt.legend()
plt.xlabel(r'Numero de cursos $X$')
plt.ylabel(r'Frecuencia $f$')
plt.title('Numero de cursos por alumno')
plt.xticks(range(2, 6 + 1))
plt.show()
```

El histograma anterior señala que los estudiantes del ICSA llevan en promedio $\bar X=4.93$ materias, con una varianza $\sigma=0.93$, con un valor aislado (*outlier*) de un estudiante que lleva solo una materia.

### La mediana se calcula de la siguiente manera usando las frecuencias $f$ ###

* ### Primero se hace la suma cumulativa de las frecuencias. Por ejemplo, si tenemos las frecuencias ###

<br>

##$$
f = [3,4,6,5,4,3,2,1]
$$##

##$$
suma\ acumulativa = [3,7,13,18,22,25,27,28]
$$##

<br>

* ### Segundo se saca la suma de todas las frecuencias $N = \sum_i f_i$###

<br>

* ### tercero, se busca el **indice** de la suma cumulativa donde se cruza el número $N/2$.



    

```python
import numpy as np
import matplotlib.pyplot as plt

# @title
f = np.array([5,9,4,3,2,4,5,3,6,0,10])  #  numero de playeras/blusa azul en tu closet
suma_cumulativa = np.cumsum(f)
N = np.sum(f)
Mediana=np.median(f)

plt.figure(figsize=(5.0,3.5))
plt.plot(np.arange(0,11),suma_cumulativa, 'sb')
plt.plot([0,11],[N/2,N/2],'--r', label = r'$N/2$')
plt.legend()
plt.title('matricula')
plt.xlabel('Indice')
plt.ylabel('Suma cumulativa')
plt.show()

sumaF=np.sum(f)
print('Suma acumulada es',suma_cumulativa)
print('Suma de las frecuencias',N)
print('la mediana es',N/2)
print('la mediana es',Mediana)
print(np.median([5,9,4,3,2,4,5,3,6,0,10]))
```

```python
# Histograma
# con grafica de distribucion de datos
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statistics as sts

# estimate kernel density:
kde=sm.nonparametric.KDEUnivariate(PSoda)
kde.fit()

plt.hist(PSoda, color='coral', edgecolor='r', alpha=0.7, density=True)
plt.plot(kde.support,kde.density, color='black', linewidth=2)
plt.xlabel('Precio del Soda')
plt.ylabel('Frecuencia')
plt.title('Distribucion de precios MATRICULA')
plt.show()

#  Calculos Estadisticos
pmedioPSoda=np.mean(PSoda)
pmedianaPSoda=np.median(PSoda)
desvstdPSoda=np.std(PSoda)
modaPSoda=sts.mode(PSoda)

print('El promedio del precio de soda es', round(pmedioPSoda,2))
print('La mediana del precio de soda es', pmedianaPSoda)
print('La desviación estandar del precio de soda es', round(desvstdPSoda,2))
print('La moda del precio de la soda es',round(modaPSoda,2))
```
