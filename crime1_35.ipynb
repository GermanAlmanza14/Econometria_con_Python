# **3.5 Historial de arresto 'CRIME1'**

* *narr86* indica el número de veces que un hombre fue arrestado en el año 1986
* *pcnv* es la proporción (no el porcentaje) de detenciones anteriores a 1986 que condujeron a una condena.
* *ptime86* es el tiempo en meses pasado en prisión durante 1986.
* *qemp86* es la cantidad de trimestres que la persona tuvo empleo en 1986.
* *avgsen* es la duración promedio de las condenas anteriores cumplidas.


<br>

```python
import statsmodels.formula.api as stf
import wooldridge as woo
import numpy as np

crime1=woo.dataWoo('crime1')
crime1
```

```python
# Estimación multilineal
regcrime1_a =stf.ols('narr86~pcnv+ptime86+qemp86', data=crime1).fit()
regcrime1_a.summary()
```

```python
0.1499*.5
```

```python
# Estimación multilineal, agregando la var 'avgsen'
regcrime1_b =stf.ols('narr86~pcnv+ptime86+qemp86+avgsen', data=crime1).fit()
regcrime1_b.summary()
```

```python
from statsmodels.iolib.summary2 import summary_col

print(summary_col([regcrime1_a,regcrime1_b],stars=True,float_format='%0.3f',
                  info_dict={'N':lambda x: "{0:d}".format(int(x.nobs)),
                             'R2':lambda x: "{:.2f}".format(x.rsquared)}))
```
